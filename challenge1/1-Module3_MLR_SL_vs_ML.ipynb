## 1. Import packages

 **Most widely used packages for Statistical Modeling and Machine Learning:**
* statsmodels → OLS, statistical modeling style (inference, p-values, CIs).
* scikit-learn → LinearRegression (OLS under the hood) and SGDRegressor (gradient descent optimization).


````python
import numpy as np
import time
from sklearn.linear_model import LinearRegression, SGDRegressor
import statsmodels.api as sm
from sklearn.preprocessing import StandardScaler
````

## 2. Make toy data

````python
### 1-Create X (features matrix) ###
# n = number of observations (rows) , p = Number of features (columns)
n, p = 50000, 1000
# ?np.random.randn
np.random.seed(0)
X = np.random.randn(n, p)
X
````

````python
### 2-Create y (Target vector) ###
true_beta = np.random.randn(p)
y = X @ true_beta + np.random.randn(n)
y
````

## 3. Model estimation

````python
### 1-Statsmodels OLS ###
# statsmodels.OLS → traditional econometrics flavor (gives summary with 𝑅2, 𝑝-values, etc.).
t0 = time.time()
X_const = sm.add_constant(X)   # add intercept
ols_model = sm.OLS(y, X_const).fit()
time_sm = time.time() - t0
time_sm
````

````python
ols_model.summary()
````

````python
# ols_model.params
````

````python
### 2-Scikit-learn LinearRegression (OLS) ###
# sklearn.LinearRegression → ML flavor but still OLS solution, optimized for prediction.
t0 = time.time()
lr = LinearRegression(fit_intercept=True)
lr.fit(X, y)
time_sklearn_lr = time.time() - t0
time_sklearn_lr
````

````python
# ?LinearRegression
````

````python
lr.intercept_
````

````python
lr.coef_
````

````python
lr.score(X,y)
````

````python
### 3-Scikit-learn SGDRegressor (Gradient Descent) ###
# sklearn.SGDRegressor → Stochastic gradient descent, useful for very large/high-dimensional data.
# Gradient Descent works best if features are on similar scales (without scaling, one feature with big values could dominate and slow convergence)

scaler = StandardScaler()                    #mean ≈ 0, variance ≈ 1
X_scaled = scaler.fit_transform(X)
t0 = time.time()
sgd = SGDRegressor(                          # Stochastic gradient descent (one sample per update) 
                   max_iter=1000,            # maximum number of passes over the training data
                   tol=1e-3,                 # stop early if loss improvement < tol
                   penalty=None,             # no regularization (pure linear regression)
                   eta0=0.1,                 # initial learning rate
                   learning_rate="constant", # learning rate stays fixed at eta0
                   random_state=0)           # reproducibility
sgd.fit(X_scaled, y)
time_sgd = time.time() - t0
time_sgd
````

````python
# ?SGDRegressor
````

````python
print(f"Statsmodels OLS time: {time_sm:.4f} sec")
print(f"Sklearn LinearRegression (OLS) time: {time_sklearn_lr:.4f} sec")
print(f"Sklearn SGDRegressor (Gradient Descent) time: {time_sgd:.4f} sec")
````
